```python
# Analyze the new memory footprint
def count_unique_parameters(model):
    """Count actual unique parameter tensors (shared params counted once)"""
    unique_params = {}
    total_params = 0
    
    for name, param in model.named_parameters():
        param_id = id(param)
        if param_id not in unique_params:
            unique_params[param_id] = {
                'tensor': param,
                'shape': param.shape,
                'elements': param.numel(),
                'first_seen': name,
                'shared_with': []
            }
        else:
            unique_params[param_id]['shared_with'].append(name)
    
    total_unique_elements = sum(info['elements'] for info in unique_params.values())
    
    return total_unique_elements, len(unique_params), unique_params

baseline_unique, baseline_objects, _ = count_unique_parameters(baseline_model)
optimized_unique, optimized_objects, optimized_params_details = count_unique_parameters(optimized_model)

baseline_state_size = sum(p.numel() for p in baseline_model.parameters())
optimized_state_size = sum(p.numel() for p in optimized_model.parameters())

print(f"Baseline model:")
print(f"   State dict size: {baseline_state_size:,} elements")
print(f"   Unique parameters: {baseline_unique:,} elements")

print(f"Optimized model:")
print(f"   State dict size: {optimized_state_size:,} elements")  # Remove the "(same as baseline)" text!
print(f"   Unique parameters: {optimized_unique:,} elements")

reduction = ((baseline_state_size - optimized_state_size) / baseline_state_size * 100)
print(f"Total parameter reduction: {reduction:.1f}%")

n_shared_params = sum([len(param_details['shared_with']) for param_id, param_details in optimized_params_details.items()])
print(f"# Shared parameters: {n_shared_params}")
```

> The analysis above showcases both parameter reduction and parameter sharing (if applied). Optimizing model parameters directly contributes to memory savings, but remember: activation memory during inference often dominates, especially with larger input sizes!

## Step 4: Initialize optimized model via transfer learning

While we have changed the architecture, we can still transfer knowledge from the trained baseline model to our optimized architecture for any matching layers.

Preserving learned features where possible reduces retraining time and maintains clinical performance.

```python
# Transfer compatible weights from baseline to optimized model
# This preserves learned features where possible, reducing retraining time

def transfer_compatible_weights(source_model, target_model):
    """
    Transfer weights from source model to target model where layer shapes match.
    This preserves learned representations in unmodified layers.
    """
    source_dict = dict(source_model.named_parameters())
    target_dict = dict(target_model.named_parameters())
    
    transferred_layers = []
    skipped_layers = []

    # Match parameters by name; the tensor `.shape` is the proxy for transferability.
    # Identical shape => the layer was left unmodified, so its learned weights are valid.
    # Different (or missing) shape => the layer was restructured by an optimization and
    # must keep its freshly initialized weights.
    with torch.no_grad():  # no autograd graph: pure in-place copy, zero memory leak
        for name, target_param in target_dict.items():
            source_param = source_dict.get(name)
            if source_param is not None and source_param.shape == target_param.shape:
                target_param.copy_(source_param)
                transferred_layers.append(name)
            else:
                skipped_layers.append(name)
    
    return transferred_layers, skipped_layers

print("Transferring compatible weights from baseline to optimized model...")
transferred, skipped = transfer_compatible_weights(baseline_model, optimized_model)

print(f"\nTransferred {len(transferred)} compatible layers")
print(f"Skipped {len(skipped)} modified/new layers")
```

### Experimental results

The table below reports the **measured** results from this notebook run. The final strategy combined `interpolation_removal` + `depthwise_separable`. Baseline figures come from Notebook 1 (sensitivity @ threshold 0.7); optimized metrics are at the selected operating threshold 0.4.

|------------|-------------------|------------|------------------|--------------------------|----------------------------|-----------------------|--------------------|-----------------|-------------|

| **Optimized** | interpolation_removal + depthwise_separable | 1,449,986 | 88.2 | 7106 | 4.19 | 0.03 | 98.5 | 96.9 | 3/5 |

> The 5 targets are: FLOPs <0.4 GFLOPs/sample, memory <100MB, single-sample latency <3ms, throughput >2000 samples/sec, sensitivity >98%. The optimized model **meets FLOPs, memory, and throughput**, but **misses single-sample latency and clinical sensitivity**.

#### Key insights and learnings

- **Massive compute and parameter reduction.** Combining native-resolution processing with depthwise separable convolutions cut FLOPs by **98.5%** (to 0.03 GFLOPs/sample) and parameters by **87%** (11,177,538 → 1,449,986). Both far exceed the targets, confirming the Notebook 1 hypothesis that the 224×224 upscaling plus dense 3×3 convolutions were the dominant costs.
- **Memory target met — and it is activation-driven.** Peak memory fell **78.9%** (418MB → 88.2MB), clearing the <100MB target. The component breakdown shows activations (14.5MB) and model parameters (5.5MB) now dominate a much smaller footprint, matching the Notebook 1 warning that activation memory scales with input resolution.
- **Throughput soared, but single-sample latency regressed.** Batched throughput improved **5.81×** (1224 → 7106 samples/sec), yet single-sample latency got *worse* (3.19ms → 4.19ms, 0.76×). Depthwise separable convolutions replace one dense kernel with two low-arithmetic-intensity kernels (depthwise + pointwise), which increases kernel-launch overhead and underutilizes the hardware at batch size 1 — great for throughput, poor for one-image latency. This is the key trade-off to resolve in the hardware phase.
- **Clinical safety was not preserved (yet).** Sensitivity dropped to **96.9%** (below the 98% floor) and accuracy fell 7.1 points to 80.1%. The aggressive depthwise factorization removed too much capacity for the short 8-epoch fine-tune to fully recover, and threshold 0.4 was not enough to lift recall back over the bar.
- **Diminishing-returns / capacity trade-off is real.** Depthwise separable delivered the parameter/FLOP wins but at a measurable accuracy cost — evidence that on this dataset the model is now capacity-limited, not compute-limited.

#### What I would do with more time (to recover the two missed targets)

- **Restore sensitivity:** train longer (15–20 epochs) with a lower LR, apply a **sensitivity-weighted loss** (class weighting / focal loss), and calibrate the decision threshold on a held-out validation split targeting recall ≥98% before locking it in. Optionally keep the first residual stage as standard convolutions (partial depthwise) to retain capacity.
- **Fix single-sample latency:** rely on the Notebook 3 hardware phase — `channels_last` + AMP, operator fusion, and TensorRT/`torch.compile` specifically address the depthwise kernel-launch overhead that hurts batch-1 latency on the T4.

### Final optimizations strategy

**Chosen strategy: `interpolation_removal` + `depthwise_separable`.** Interpolation removal is the mandatory, root-cause fix from Notebook 1 and by itself drives most of the FLOP/memory/throughput gains at essentially no accuracy risk. Depthwise separable convolutions add the large parameter/FLOP reduction that pushes throughput to 5.81× and parameters down 87%. I deliberately left the more exotic techniques (grouped conv, inverted residuals, low-rank, parameter sharing) **off** because the model is already compute-light and further compression only deepens the sensitivity deficit. Channel optimization (`channels_last` + in-place ReLU) is deferred to Notebook 3 where the hardware backend can actually exploit it.

**Overall Production Readiness**: **3/5 targets achieved → Needs Additional Work (efficiency-ready, clinical-safety pending)**

The efficiency profile is production-grade — FLOPs, memory, and throughput all clear their targets with wide margins. The two open items are (1) recovering sensitivity to ≥98% via sensitivity-focused retraining and threshold calibration, and (2) closing the single-sample latency gap, which the Notebook 3 hardware acceleration (AMP, fusion, TensorRT) is specifically designed to address. Given the enormous compute headroom already achieved, both gaps are recoverable without sacrificing the efficiency wins.

### Quiz question: Multi-label classification adaptation

**How would your optimization strategy change if the task was multi-label classification from chest X-rays (as in [ChestMNIST](https://github.com/rsm-13/classifying-chestMNIST) which has 14 labels including cardiomegaly, mass, pneumonia, ...) instead of binary classification?**

Multi-label classification changes the head and the decision logic more than the backbone, but it raises the stakes for aggressive compression — and our own result here (an 87% parameter cut costing 2.6 points of sensitivity) is a direct warning of that risk:

- **Output head and loss.** The 2-way softmax head becomes 14 independent sigmoid outputs trained with `BCEWithLogitsLoss` (often class-weighted for imbalance). Each label needs its **own** operating threshold, so a single global threshold no longer works — threshold selection becomes per-pathology.
- **Capacity matters more.** 14 co-occurring pathologies with subtle, localized features (a small mass vs. cardiomegaly) demand richer representations. Given that depthwise separation already cost us sensitivity on a *binary* task, I would apply the most compressive techniques (low-rank factorization, parameter sharing, aggressive depthwise) far more conservatively and re-validate **per-label** sensitivity, not just aggregate accuracy.
- **Resolution trade-off shifts.** Interpolation removal is still valuable, but small findings may need more spatial detail than 64×64 provides, so I might process at a higher native resolution (e.g., 128×128) — trading some efficiency for diagnostic recall on small lesions.
- **Evaluation and safety.** Metrics move to per-class AUC / macro-AUC and mean AP; "sensitivity >98%" must be checked **per critical label**, since an optimization can preserve overall accuracy while quietly degrading a rare but dangerous class.

In short: keep interpolation removal (low-risk, hardware-friendly), be much more cautious with parameter-reducing techniques given the sensitivity hit we already observed, possibly raise the native resolution, and drive the whole process with per-label clinical metrics rather than a single binary threshold.
